* **S2.T2.3. Para las variables que tengan menos de 10 categorias distintas, representad gráficamente cómo se distribuyen las observaciones. Seleccione y represente gráficamente 5 variables que ilustren distintos escenarios de distribución. El objetivo es identificar casos que condicionen el valor de la información, tales como:**

  * ***Variables sin variabilidad:*** Una sola categoría es absoluta (valor informativo nulo o casi nulo).
  * ***Variables equilibradas:*** Repartición homogénea entre categorías.
  * ***Variables con baja calidad:*** El volumen de valores nulos (NaN) compromete la utilidad de la columna.
  * ***Variables sesgadas:*** Grandes disparidades de frecuencia entre los grupos observados.
  * ***Variables bimodales:*** Existencia de dos grupos dominantes que dividen la muestra en dos realidades.
  * ***Variables de bajo peso (Ruido):*** Muchas categorías con frecuencias muy bajas que no permiten extraer patrones.

```python
# Selección de 5 variables representativas de distintos escenarios
variables_seleccionadas = [
    "NOM_PROVINCIA", # Variable 1
    "ZONA CLIMATICA", # Variable 2
    "Qualificació de consum d'energia primaria no renovable", # Variable 3
    "Qualificacio d'emissions de CO2", # Variable 4
    "SOLAR TERMICA"  # Variable 5
]

# Creamos una figura con una cuadrícula de 3 filas y 2 columnas
fig, axes = plt.subplots(nrows=3, ncols=2, figsize=(15, 12))
axes = axes.flatten() # Aplanamos la matriz de ejes para iterar fácilmente

for i, col in enumerate(variables_seleccionadas):
    if col:
        # Dibujamos cada gráfico en su posición correspondiente (ax)
        df[col].value_counts(dropna=False).plot(
            kind="bar",
            ax=axes[i],
            color="slategray",
            edgecolor="black"
        )
        axes[i].set_title(f"Distribución: {col}", fontsize=12, fontweight='bold')
        axes[i].set_ylabel("Nº de observaciones")
        axes[i].tick_params(axis='x', rotation=45)

# Eliminamos el último gráfico (el sexto) que quedará vacío
fig.delaxes(axes[5])

# Ajustamos el espaciado para que no se solapen los títulos
plt.tight_layout()
plt.show()
```

  * **Variabilidad Nula:** Variables que presentan el mismo valor (o casi el mismo) para toda la muestra.
  * **Outliers (Valores Atípicos):** Datos que se sitúan fuera del rango físicamente lógico (ej. años de construcción futuros o consumos imposibles).
  * **Distribuciones Multimodales:** Presencia de varios "picos", lo que sugiere que la muestra mezcla tipologías de edificios muy distintas.
  * **Acumulación en el Cero:** Gran concentración de valores en 0 que podría indicar datos faltantes mal registrados.
